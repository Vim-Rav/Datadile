## 1. Business Context/Story

### The Business Scenario

Northstar Commerce is a growing online retail company. The company sells lifestyle products through its website and acquires customers through several marketing channels:

* Paid search advertising
* Paid social media campaigns
* Email marketing
* Organic search
* Referral campaigns
* Direct website visits

Over the past six months, the marketing team has increased its advertising expenditure. However, revenue has not grown consistently with spending.

Northstar's marketing manager suspects that some campaigns are generating large numbers of website visitors but very few purchases. Other campaigns appear to attract fewer visitors but generate more valuable customers.

Unfortunately, the company tracks advertising performance, website activity, and sales in separate spreadsheets. The marketing team cannot easily connect advertising expenditure to customer purchases or determine which customer groups generate the most revenue.


#### Northstar's request

>"We are spending more on digital marketing, but we don't know which channels are actually generating profitable customers. We need a clear picture of our marketing performance, customer purchasing behaviour, and sales trends.
>
>Please analyse our marketing and transaction data, identify the campaigns and customer segments that offer the greatest opportunities, and build a dashboard that allows our team to monitor performance without manually combining spreadsheets every week.
>
>We also want practical recommendations on how to improve conversion rates and allocate our marketing budget more effectively."

### The business questions

The analysis must answer these questions:

Marketing performance
* Which marketing channels generate the most revenue?
* Which channels deliver the lowest customer acquisition cost?
* Which campaigns produce the strongest return on advertising spend?
* Are there campaigns that consume significant budget without generating proportional revenue?

Customer behaviour
* Which customer segments generate the highest revenue?
* How many customers make repeat purchases?
* What is the average order value for different customer groups?
* Which customer groups show signs of declining engagement?

Sales performance
* How do sales vary by month, product category, and acquisition channel?
* Are sales growing because the company is acquiring more customers or because existing customers are spending more?

Business optimization
* Where are the biggest opportunities to improve the marketing funnel?
* How should the company prioritize its marketing investment?

Important analytical principle

Marketing attribution can be complicated. A customer might click an advertisement, visit through organic search later, and then purchase after receiving an email.

You will distinguish between directly observed performance and attributed performance. You will not claim that a channel caused a sale simply because the two are associated.

Similarly, ROAS is not the same as profitability. A campaign can generate substantial revenue while losing money after product costs, fulfillment, and advertising expenditure.

This distinction will make the project more credible to employers and freelance clients.
